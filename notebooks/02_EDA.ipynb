```sql
select passenger_count ,count(*) as total_trips from nyc_taxi.bronze.yellow_trip_raw group by passenger_count order by total_trips desc;
```

```sql

select max(trip_distance) as max_trip_distance,
min(trip_distance) as min_trip_distance,
MAX(fare_amount) AS MAX_FARE,
MIN(fare_amount) AS MIN_FARE,
MAX(total_amount) AS MAX_TOTAL,
MIN(total_amount) AS MIN_TOTAL from nyc_taxi.bronze.yellow_trip_raw;

```

```sql
SELECT AVG(TRIP_DISTANCE) AS AVG_TRIP_DISTANCE,
AVG(FARE_AMOUNT) AS AVG_FARE,
AVG(tolls_amount) AS AVG_TOLLS,
AVG(total_amount) AS AVG_TOTAL
 FROM nyc_taxi.bronze.yellow_trip_raw;
```

```sql
SELECT PAYMENT_TYPE ,count(*) as trips from nyc_taxi.bronze.yellow_trip_raw group by payment_type order by trips desc;
```

```sql
select hour(tpep_pickup_datetime) ,count(*) as peak_hours_count from nyc_taxi.bronze.yellow_trip_raw group by hour(tpep_pickup_datetime) order by peak_hours_count desc; 
```
