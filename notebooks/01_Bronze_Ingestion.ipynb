```sql
CREATE CATALOG if not exists nyc_taxi;
CREATE SCHEMA  if not exists nyc_taxi.raw;

CREATE VOLUME if not exists  nyc_taxi.raw.trip_data;
```

```sql
%python
df = spark.read.parquet(
    "/Volumes/nyc_taxi/raw/trip_data/"
)
df.printSchema()
print("Total rows:",df.count())
print("Total Columns:",len(df.columns))
```

```sql
CREATE SCHEMA IF NOT EXISTS nyc_taxi.bronze;
```

```sql
%python
df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("nyc_taxi.bronze.yellow_trip_raw")
```

```sql
select * from nyc_taxi.bronze.yellow_trip_raw;
```

```sql
select count(*) as total_rows from nyc_taxi.bronze.yellow_trip_raw;
```

```sql
SELECT
    SUM(CASE WHEN passenger_count IS NULL THEN 1 ELSE 0 END) AS passenger_count_nulls,
    SUM(CASE WHEN trip_distance IS NULL THEN 1 ELSE 0 END) AS trip_distance_nulls,
    SUM(CASE WHEN fare_amount IS NULL THEN 1 ELSE 0 END) AS fare_amount_nulls,
    SUM(CASE WHEN total_amount IS NULL THEN 1 ELSE 0 END) AS total_amount_nulls
FROM nyc_taxi.bronze.yellow_trip_raw;
```
