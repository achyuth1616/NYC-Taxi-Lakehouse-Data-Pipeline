```sql
select round(sum(total_amount),2) as total_revenue from nyc_taxi.silver.yellow_trip_clean ;
```

```sql
select count(*) as total_trips from nyc_taxi.silver.yellow_trip_clean;
```

```sql
SELECT round(avg(fare_amount),2) as average_fare_amount from nyc_taxi.silver.yellow_trip_clean;
```

```sql
select round(avg(trip_distance),2) as average_trip_distance from nyc_taxi.silver.yellow_trip_clean;

```

```sql
select sum(passenger_count) as total_passengers from nyc_taxi.silver.yellow_trip_clean;
```

```sql
select payment_type,round(sum(total_amount),2) as total_revenue from nyc_taxi.silver.yellow_trip_clean group by payment_type order by total_revenue desc;
```

```sql
select PULocationID,count(*) as total_trips from nyc_taxi.silver.yellow_trip_clean group by pulocationid order by total_trips desc limit 10;
```

```sql
select pulocationid,round(sum(total_amount),2) as total_revenue from nyc_taxi.silver.yellow_trip_clean group by PULocationID order by total_revenue desc limit 10 ;
```

```sql
select hour(tpep_pickup_datetime) as hour,count(*) as total_trips from nyc_taxi.silver.yellow_trip_clean group by hour(tpep_pickup_datetime) order by total_trips desc  limit  5;
```
